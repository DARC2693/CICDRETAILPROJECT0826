```python
# =====================================================================
# proceso / 00_prepamb.py
# Task "prepamb" del Job (ver Ejemplo 2). Ejecuta PrepAmb/01_prepamb.sql
# sustituyendo ${catalogo} y ${storageName} por los valores reales del
# ambiente (dev/prod), igual al patrón de widgets visto en clase.
# =====================================================================
```

```python
dbutils.widgets.removeAll()
```

```python
dbutils.widgets.text("sql_file", "../PrepAmb/01_prepamb.sql")
dbutils.widgets.text("catalogo", "retail_medallion")
dbutils.widgets.text("storageName", "adlsretailproject0826")

sql_file = dbutils.widgets.get("sql_file")
catalogo = dbutils.widgets.get("catalogo")
storageName = dbutils.widgets.get("storageName")
```

```python
with open(sql_file, "r", encoding="utf-8") as f:
    script = f.read()

script = script.replace("${catalogo}", catalogo).replace("${storageName}", storageName)

statements = [s.strip() for s in script.split(";") if s.strip() and not s.strip().startswith("--")]

for stmt in statements:
    print(f"Ejecutando:\n{stmt}\n")
    spark.sql(stmt)

print(f"Ambiente preparado: {len(statements)} sentencias ejecutadas sobre catalog '{catalogo}'.")
```
